```python
# Q.(19)Create a random 4x4 matrix and use NumPy to compute itseigenvalues and eigenvectors. Verify the eigenvalues by reconstructing thematrix. 
 
# 1. Create a random 4x4 matrix
A = np.random.rand(4, 4)

# 2. Compute eigenvalues and eigenvectors
# eigenvalues (w) and eigenvectors (v)
eigenvalues, eigenvectors = np.linalg.eig(A)

# 3. Verify the result (Reconstruct the matrix)
# Formula: A = V * diag(W) * V_inv
V = eigenvectors
W_diag = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)

A_reconstructed = V @ W_diag @ V_inv

# Check if the original and reconstructed matrices match
print("Original Matrix matches Reconstructed:", np.allclose(A, A_reconstructed))  
```

```python
# Q.(27)Generate a large 2D array and demonstrate the use of np. reshape() and unravel () to manipulate its shape for various linear 
# algebra operations. 

import numpy as np

# 1. Generate a large 2D array (100 rows, 100 columns) with random numbers
np.random.seed(42)
large_2d = np.random.randn(100, 100)
print(f"Original 2D Array Shape: {large_2d.shape}")

# --- Using np.reshape() for Linear Algebra ---
# Suppose a linear algebra function expects a vector instead of a matrix.
# We reshape the 10,000 elements into a 1D column vector.
vector = np.reshape(large_2d, (10000, 1))
print(f"Reshaped to Column Vector Shape: {vector.shape}")

# Or we reshape it into a 3D tensor for batch matrix multiplications (e.g., 4 batches of 50x50 matrices)
tensor_3d = np.reshape(large_2d, (4, 50, 50))
print(f"Reshaped for Batch Operations Shape: {tensor_3d.shape}\n")

# --- Using np.unravel_index() to Find Matrix Coordinates ---
# Let's say we find the absolute maximum value in the entire matrix.
# np.argmax() flattens the array by default and gives us a single "flat index".
flat_max_idx = np.argmax(large_2d)
print(f"Flat index of maximum value: {flat_max_idx}")

# Linear algebra operations often require the exact row and column (matrix coordinates).
# np.unravel_index translates that 1D flat index back into the 2D (row, col) coordinates.
row, col = np.unravel_index(flat_max_idx, large_2d.shape)

print(f"Unraveled Matrix Coordinates: Row {row}, Column {col}")
print(f"Verified Maximum Value: {large_2d[row, col]:.4f}")
```

```python
# Q.(28)Given a 6x6 matrix, use advanced indexing and slicing to extractthe upper triangular part of the matrix and set the lower triangular part to 
# zero. Verify the result. 

import numpy as np

# 1. Create a 6x6 matrix with sequential integers
matrix = np.arange(1, 37).reshape(6, 6)
print("Original 6x6 Matrix:\n", matrix, "\n")

# 2. Extract the upper triangular part
# np.triu_indices(6) returns the exact row and column coordinates for the upper triangle
upper_rows, upper_cols = np.triu_indices(6)
upper_triangular_elements = matrix[upper_rows, upper_cols]

print("Extracted Upper Triangular Elements:\n", upper_triangular_elements, "\n")

# 3. Set the lower triangular part to zero
# k=-1 excludes the main diagonal, so we only target the elements below it
lower_rows, lower_cols = np.tril_indices(6, k=-1)
matrix[lower_rows, lower_cols] = 0

# 4. Verify the result
print("Verified Modified Matrix (Lower Triangle Zeroed):\n", matrix)
```

```python
                                            # Visualization with Matplotlib and Seaborn #

# Q.(1) Bar chart showing the total deaths per year for 1984-2016 

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load your dataset
df = pd.read_csv("U.S.Death.csv")
df
# 2. Filter data for the years 1984 to 2016
# (Note: Standard datasets usually name this column 'year' or 'Year')
df_filtered = df[(df["year"] >= 1984) & (df["year"] <= 2016)]

# 3. Group by year and calculate total deaths
deaths_by_year = df_filtered.groupby("year").size().reset_index(name="Deaths")

# 4. Set up the plotting dimensions
plt.figure(figsize=(15, 6))
sns.set_theme(style="whitegrid")

# 5. Create the bar chart
ax = sns.barplot(
    data=deaths_by_year, x="year", y="Deaths", color="midnightblue"
)

# 6. Customize labels, title, and ticks
plt.title(
    "Total U.S. Police Deaths per Year (1984 - 2016)",
    fontsize=16,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Number of Deaths", fontsize=12)
plt.xticks(rotation=45)  # Tilts the years so they don't overlap

# 7. Add data labels on top of each bar for a cleaner look
for p in ax.patches:
    ax.annotate(
        format(int(p.get_height()), "d"),
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 5),
        textcoords="offset points",
        fontsize=9,
    )

# Adjust layout and show the plot
plt.tight_layout()
plt.show()
```

```python
# Q.(2) Line chart comparing the yearly deaths in different states 

import matplotlib.pyplot as plt
import pandas as pd

# 1. Load the data
df = pd.read_csv("U.S.Death.csv")

# 2. Clean up any hidden spaces in the state names
df["state"] = df["state"].str.strip()

# 3. Filter for years between 1984 and 2016
filtered_df = df[(df["year"] >= 1984) & (df["year"] <= 2016)]

# 4. Choose a few large states to compare so the graph stays clean
states_to_compare = filtered_df[filtered_df["state"].isin(["TX", "CA", "NY"])]

# 5. Count the number of deaths per year for each chosen state
state_yearly_counts = (
    states_to_compare.groupby(["year", "state"]).size().unstack(fill_value=0)
)

# 6. Plot the data as a line chart with small dots (markers)
state_yearly_counts.plot(kind="line", marker="o", figsize=(12, 5))

# 7. Add titles and labels
plt.title("Yearly Police Deaths Comparison by State (1984-2016)")
plt.xlabel("Year")
plt.ylabel("Number of Deaths")

# 8. Show the graph
plt.show()
```

```python
# Q.(9) A tree map with three levels: state, city (description), cause

import pandas as pd
import plotly.express as px

# 1. Load the data
df = pd.read_csv("U.S.Death.csv")

# 2. Clean up hidden blank spaces in the state names
df["state"] = df["state"].str.strip()

# 3. Filter for years between 1984 and 2016
filtered_df = df[(df["year"] >= 1984) & (df["year"] <= 2016)]

# 4. Group data to calculate total counts for each level
# This keeps the map performance fast and organized
treemap_data = (
    filtered_df.groupby(["state", "dept_name", "cause_short"])
    .size()
    .reset_index(name="Total Deaths")
)

# 5. Build the three-level Tree Map
fig = px.treemap(
    treemap_data,
    path=[
        "state",
        "dept_name",
        "cause_short",
    ],  # Defines the 3 hierarchical levels
    values="Total Deaths",  # Determines rectangle size based on total deaths
    color="state",  # Gives each state a unique shade
    title="Multi-Level Tree Map of U.S. Police Deaths (1984-2016)",
)

# 6. Adjust map formatting
fig.update_layout(margin=dict(t=50, l=25, r=25, b=25))

# 7. Display the interactive chart
fig.show()
```

```python
# Q.(3) Create a function to automate downloading for differenttopics/search queries. 

import os
import requests


def download_search_results(query):
    """Automates downloading and saving a webpage based on a search query."""
    # 1. Define the base search URL (Example: a practice site or a search endpoint)
    # For this example, we'll use a public mock/search structure
    base_url = "https://httpbin.org/anything"  # Generates a mock response of what we sent

    # Alternatively, if you were scraping a real search engine or site:
    # base_url = "https://html.duckduckgo.com/html/"

    # 2. Define the search parameters
    # The dictionary key (like 'q') depends on what the target website uses
    search_parameters = {"q": query}

    # 3. Set headers to look like a real browser browser
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
    }

    try:
        print(f"🔄 Automating download for query: '{query}'...")

        # 4. Send the GET request with the query parameters attached
        response = requests.get(
            base_url, params=search_parameters, headers=headers, timeout=10
        )

        if response.status_code == 200:
            # 5. Create a safe, dynamic filename based on the search query
            # We replace spaces with underscores to keep filenames clean
            clean_filename = f"search_{query.replace(' ', '_').lower()}.html"

            # 6. Save the file locally
            with open(clean_filename, "w", encoding="utf-8") as file:
                file.write(response.text)

            print(f"✅ Success! Saved locally as: '{clean_filename}'\n")
        else:
            print(
                f"❌ Failed for '{query}'. Status code: {response.status_code}\n"
            )

    except requests.exceptions.RequestException as e:
        print(f"💥 Network error occurred for '{query}': {e}\n")


# --- Testing the Automation ---
# Define a list of different topics you want to download
topics_to_scrape = ["python programming", "data science", "web scraping"]

# Loop through the topics and let the function do the work automatically
for topic in topics_to_scrape:
    download_search_results(topic)
```

```python
# Q.(7) Create functions to extract from the page into lists and dictionaries. 

from bs4 import BeautifulSoup

# Load your local file
filename = "downloaded_page.html"

with open(filename, "r", encoding="utf-8") as file:
    soup = BeautifulSoup(file.read(), "html.parser")


# --- FUNCTION 1: Extract into a Simple List ---
def extract_top_tags(soup_object):
    """Extracts the trending tags on the page into a clean Python list of strings."""
    tags_list = []
    
    # Locate the container box holding the trending tags
    tags_container = soup_object.find("div", class_="tags-box")
    if tags_container:
        # Find all individual anchor elements inside it
        tag_elements = tags_container.find_all("a", class_="tag")
        for tag in tag_elements:
            tags_list.append(tag.text.strip())
            
    return tags_list


# --- FUNCTION 2: Extract into a List of Dictionaries ---
def extract_structured_quotes(soup_object):
    """Extracts all quotes on the page into a structured list of dictionaries."""
    all_quotes_data = []
    
    # Locate all individual quote container cards
    quote_blocks = soup_object.find_all("div", class_="quote")
    
    for block in quote_blocks:
        # Extract the core text, author name, and profile path
        text = block.find("span", class_="text").text.strip()
        author = block.find("small", class_="author").text.strip()
        author_url = block.find("a")["href"]
        
        # Extract all sub-tags associated with THIS specific quote card into a mini-list
        quote_tags = [t.text.strip() for t in block.find_all("a", class_="tag")]
        
        # Package the structured properties of this item into a dictionary record
        quote_dictionary = {
            "quote_text": text,
            "author_name": author,
            "author_bio_link": f"https://quotes.toscrape.com{author_url}",
            "associated_tags": quote_tags
        }
        
        # Append the individual item dictionary to our primary master list
        all_quotes_data.append(quote_dictionary)
        
    return all_quotes_data


# --- Executing and Testing the Functions ---
print("⚙️ RUNNING DATA STRUCTURE EXTRACTION FUNCTIONS ⚙️\n" + "="*60)

# Call Function 1
trending_tags = extract_top_tags(soup)
print(f"📋 RESULT FROM FUNCTION 1 (Simple List):\n{trending_tags}\n")
print("-" * 60)

# Call Function 2
extracted_dataset = extract_structured_quotes(soup)
print(f"🗃️ RESULT FROM FUNCTION 2 (List of Dictionaries - First 2 Records):\n")

# Displaying just the first 2 dictionary items cleanly for readability
for item in extracted_dataset[:2]:
    print(item)
    print()
```

```python
# Q.(8) Create functions for the end-to-end process of downloading, parsing,and saving CSVs.

import csv
import os
import pandas as pd
import requests
from bs4 import BeautifulSoup


# --- STEP 1: Automated Downloading ---
def download_webpage(url, output_html_path):
    """Downloads a webpage and saves it locally to cache it."""
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
    }
    try:
        response = requests.get(url, headers=headers, timeout=10)
        if response.status_code == 200:
            with open(output_html_path, "w", encoding="utf-8") as file:
                file.write(response.text)
            print(
                f"✅ Webpage downloaded successfully and cached at: '{output_html_path}'"
            )
            return True
        else:
            print(
                f"❌ Failed to download. HTTP Status Code: {response.status_code}"
            )
            return False
    except requests.exceptions.RequestException as e:
        print(f"💥 Network Error: {e}")
        return False


# --- STEP 2: Parsing & Information Extraction ---
def parse_local_html(html_path):
    """Parses the cached local HTML file and transforms it into a list of dictionaries."""
    if not os.path.exists(html_path):
        print(f"❌ Error: Cache file '{html_path}' not found.")
        return []

    with open(html_path, "r", encoding="utf-8") as file:
        soup = BeautifulSoup(file.read(), "html.parser")

    quotes_dataset = []
    quote_blocks = soup.find_all("div", class_="quote")

    for index, block in enumerate(quote_blocks, 1):
        text_element = block.find("span", class_="text")
        author_element = block.find("small", class_="author")
        tag_elements = block.find_all("a", class_="tag")

        # Compile individual tag words into a comma-separated string for structured column mapping
        tags_list = [tag.text.strip() for tag in tag_elements]

        # Structure elements into a clean row data record
        record = {
            "ID": index,
            "Quote": text_element.text.strip() if text_element else "",
            "Author": author_element.text.strip() if author_element else "",
            "Tags": ", ".join(tags_list),
        }
        quotes_dataset.append(record)

    print(
        f"🔍 Parsing Complete: Successfully extracted {len(quotes_dataset)} records."
    )
    return quotes_dataset


# --- STEP 3: Structural Saving ---
def save_to_csv(data_list, output_csv_path):
    """Saves a structured list of records into a clean tabular CSV spreadsheet."""
    if not data_list:
        print("⚠️ No data data records found to export.")
        return False

    # Load data into a Pandas DataFrame for validation and quick processing
    df = pd.DataFrame(data_list)

    # Save to file path without adding a redundant pandas index column
    df.to_csv(output_csv_path, index=False, encoding="utf-8")
    print(
        f"💾 Data Pipeline Complete: Exported spreadsheet to '{output_csv_path}'"
    )
    return True


# --- PIPELINE ORCHESTRATOR ---
def run_end_to_end_pipeline(target_url, cache_html, final_csv):
    """Orchestrates the data scraping pipeline sequentially from start to finish."""
    print("🚀 STARTING END-TO-END AUTOMATED PIPELINE...")
    print("=" * 65)

    # Execution flow sequence
    if download_webpage(target_url, cache_html):
        extracted_data = parse_local_html(cache_html)
        if save_to_csv(extracted_data, final_csv):
            print("=" * 65)
            print("🎉 ALL TASKS COMPLETED! DATA CLEANED AND EXPORTED.")


# --- Executing the Automated Process ---
target_website = "https://quotes.toscrape.com/"
cached_file = "downloaded_page.html"
export_file = "scraped_quotes_dataset.csv"

run_end_to_end_pipeline(target_website, cached_file, export_file)
```
