```python
                                    # Clinic Appointment System #
class ClinicAppointment:
    def __init__(self):
        # Dictionary to store bookings: {mobile: {details}}
        self.appointments = {}
        # List of available slots
        self.slots = ["10am", "11am", "12pm", "2pm", "3pm"]
        # Track slot counts per doctor: {(doctor, slot): count}
        self.doctor_load = {}

    def book_appointment(self):
        print("\n--- Book Appointment ---")
        name = input("Enter Patient Name: ")
        age = input("Enter Patient Age: ")
        mobile = input("Enter Mobile Number: ")
        doctor = input("Enter Preferred Doctor: ")

        print(f"Available Slots: {', '.join(self.slots)}")
        slot = input("Choose a slot: ")

        if slot not in self.slots:
            print("Invalid slot selection.")
            return

        # Check availability (Max 3 appointments per slot per doctor)
        current_load = self.doctor_load.get((doctor, slot), 0)
        
        if current_load < 3:
            # Store appointment data
            self.appointments[mobile] = {
                "name": name,
                "age": age,
                "doctor": doctor,
                "slot": slot
            }
            # Update doctor load
            self.doctor_load[(doctor, slot)] = current_load + 1
            print(f"Success! Appointment confirmed for {name} with Dr. {doctor} at {slot}.")
        else:
            print(f"Sorry, Dr. {doctor} is fully booked for {slot}. Try another time.")

    def view_appointment(self):
        mobile = input("\nEnter your registered Mobile Number to view: ")
        if mobile in self.appointments:
            app = self.appointments[mobile]
            print("\n--- Your Appointment Details ---")
            print(f"Name: {app['name']}\nDoctor: {app['doctor']}\nTime: {app['slot']}")
        else:
            print("No appointment found for this number.")

    def cancel_appointment(self):
        mobile = input("\nEnter Mobile Number to cancel: ")
        if mobile in self.appointments:
            # Get details to reduce doctor load
            app = self.appointments.pop(mobile)
            self.doctor_load[(app['doctor'], app['slot'])] -= 1
            print("Appointment cancelled successfully.")
        else:
            print("Record not found.")

# --- Execution ---
clinic = ClinicAppointment()

# Simple menu loop
while True:
    print("\n1. Book | 2. View | 3. Cancel | 4. Exit")
    choice = input("Select an option: ")
    
    if choice == '1': clinic.book_appointment()
    elif choice == '2': clinic.view_appointment()
    elif choice == '3': clinic.cancel_appointment()
    elif choice == '4': break
    else: print("Invalid choice.")
```

```python
                                    # School Management System #

class SchoolManagement:
    def __init__(self):
        # Dictionary to store student records: {student_id: {details}}
        self.students = {}
        # Counter to generate unique Student IDs starting from 1001
        self.next_id = 1001

    def new_admission(self):
        print("\n--- New Admission ---")
        name = input("Enter Student Name: ")
        
        # Validate Age (5-18)
        try:
            age = int(input("Enter Age (5-18): "))
            if not (5 <= age <= 18):
                print("Admission Denied: Age must be between 5 and 18.")
                return
        except ValueError:
            print("Invalid input! Please enter a number for age.")
            return

        # Validate Class (1-12)
        grade = input("Enter Class (1-12): ")
        
        # Validate Mobile Number (10 digits)
        mobile = input("Enter Guardian's 10-digit Mobile Number: ")
        if not (mobile.isdigit() and len(mobile) == 10):
            print("Invalid Mobile Number! It must be exactly 10 digits.")
            return

        # Assign ID and Save
        student_id = self.next_id
        self.students[student_id] = {
            "name": name,
            "age": age,
            "class": grade,
            "mobile": mobile
        }
        
        print(f"Admission Successful! Student ID: {student_id}")
        self.next_id += 1  # Increment for the next student

    def view_student(self):
        try:
            sid = int(input("\nEnter Student ID to search: "))
            if sid in self.students:
                s = self.students[sid]
                print(f"\n--- Record for ID {sid} ---")
                print(f"Name: {s['name']}\nAge: {s['age']}\nClass: {s['class']}\nMobile: {s['mobile']}")
            else:
                print("Student ID not found.")
        except ValueError:
            print("Please enter a valid numeric ID.")

    def update_student(self):
        try:
            sid = int(input("\nEnter Student ID to update: "))
            if sid in self.students:
                print("1. Update Mobile | 2. Update Class")
                choice = input("Choice: ")
                if choice == '1':
                    new_mobile = input("Enter new 10-digit mobile: ")
                    if len(new_mobile) == 10 and new_mobile.isdigit():
                        self.students[sid]['mobile'] = new_mobile
                        print("Mobile updated.")
                    else:
                        print("Invalid number.")
                elif choice == '2':
                    self.students[sid]['class'] = input("Enter new Class: ")
                    print("Class updated.")
            else:
                print("ID not found.")
        except ValueError:
            print("Invalid input.")

    def remove_student(self):
        try:
            sid = int(input("\nEnter Student ID to remove: "))
            if sid in self.students:
                del self.students[sid]
                print(f"Student ID {sid} removed from records.")
            else:
                print("ID not found.")
        except ValueError:
            print("Invalid input.")

# --- Menu Logic ---
school = SchoolManagement()

while True:
    print("\n=== School Management Menu ===")
    print("1. New Admission\n2. View Student\n3. Update Student\n4. Remove Student\n5. Exit")
    choice = input("Select an option: ")

    if choice == '1': school.new_admission()
    elif choice == '2': school.view_student()
    elif choice == '3': school.update_student()
    elif choice == '4': school.remove_student()
    elif choice == '5': 
        print("Exiting system. Goodbye!")
        break
    else: 
        print("Invalid choice, try again.")
```

```python
                                        # Transport Reservation System (Bus Ticketing) #

import random

class BusReservation:
    def __init__(self):
        # Predefined routes and prices
        self.routes = {
            "1": {"path": "Mumbai to Pune", "price": 500},
            "2": {"path": "Delhi to Jaipur", "price": 600},
            "3": {"path": "Ahmedabad to Udaipur", "price": 700},
            "4": {"path": "Bangalore to Chennai", "price": 800}
        }
        # Track filled seats per route: {route_id: seat_count}
        self.route_seats = {"1": 0, "2": 0, "3": 0, "4": 0}
        # Store tickets: {ticket_id: {details}}
        self.tickets = {}

    def show_routes(self):
        print("\n--- Available Routes ---")
        for r_id, info in self.routes.items():
            print(f"{r_id}. {info['path']} - ₹{info['price']}")

    def book_ticket(self):
        self.show_routes()
        route_choice = input("\nSelect Route ID to book: ")

        if route_choice not in self.routes:
            print("Invalid Route!")
            return

        # Check Seat Availability (Max 40)
        if self.route_seats[route_choice] >= 40:
            print("Sorry, this bus is full!")
            return

        name = input("Enter Passenger Name: ")
        age = input("Enter Age: ")
        mobile = input("Enter Mobile Number: ")

        # Generate Unique Ticket ID and Seat Number
        ticket_id = f"TIC{random.randint(1000, 9999)}"
        self.route_seats[route_choice] += 1
        seat_no = self.route_seats[route_choice]

        # Save Ticket
        self.tickets[ticket_id] = {
            "name": name,
            "age": age,
            "mobile": mobile,
            "route": self.routes[route_choice]['path'],
            "price": self.routes[route_choice]['price'],
            "seat": seat_no,
            "route_id": route_choice
        }

        print(f"\nBooking Confirmed! ID: {ticket_id} | Seat No: {seat_no}")

    def view_ticket(self):
        tid = input("\nEnter Ticket ID: ")
        if tid in self.tickets:
            t = self.tickets[tid]
            print("\n--- Ticket Details ---")
            print(f"Passenger: {t['name']}\nRoute: {t['route']}\nSeat: {t['seat']}\nPrice: ₹{t['price']}")
        else:
            print("Ticket ID not found.")

    def cancel_ticket(self):
        tid = input("\nEnter Ticket ID to cancel: ")
        if tid in self.tickets:
            # Reduce the seat count for that route before deleting
            ticket_info = self.tickets.pop(tid)
            self.route_seats[ticket_info['route_id']] -= 1
            print(f"Ticket {tid} has been cancelled successfully.")
        else:
            print("Invalid Ticket ID.")

# --- Menu Loop ---
bus_system = BusReservation()

while True:
    print("\n=== Transport Reservation System ===")
    print("1. Show Routes\n2. Book Ticket\n3. View Ticket\n4. Cancel Ticket\n5. Exit")
    choice = input("Select an option: ")

    if choice == '1': bus_system.show_routes()
    elif choice == '2': bus_system.book_ticket()
    elif choice == '3': bus_system.view_ticket()
    elif choice == '4': bus_system.cancel_ticket()
    elif choice == '5': break
    else: print("Invalid choice.")
```
