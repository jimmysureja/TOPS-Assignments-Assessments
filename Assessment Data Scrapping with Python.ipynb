```python
                                                    # Data Scrapping with Python #

import requests
from bs4 import BeautifulSoup

# Step 1: Define the URL you want to scrape
url = "http://quotes.toscrape.com"

# Step 2: Fetch the web page contents using requests
response = requests.get(url)

# Step 3: Parse the content and generate the 'soup' object
# 'response.text' contains the raw HTML code of the page
soup = BeautifulSoup(response.text, 'html.parser')

print("--- Soup Object Created Successfully ---\n")

# Step 4: Use selectors to find elements and extract information
# Every quote on this website is contained inside a <div> with the class 'quote'
quote_elements = soup.find_all('div', class_='quote')

# Step 5: Loop through the elements to extract text and author names
for quote in quote_elements:
    # Extract the text of the quote (found in a <span> with class 'text')
    text = quote.find('span', class_='text').text
    
    # Extract the author name (found in a <small> tag with class 'author')
    author = quote.find('small', class_='author').text
    
    # Print the extracted information cleanly
    print(f"Quote: {text}")
    print(f"Author: {author}")
    print("-" * 40)
```
