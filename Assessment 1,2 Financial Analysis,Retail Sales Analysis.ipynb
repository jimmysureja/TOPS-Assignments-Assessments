```python
import pandas as pd 
fin = pd.read_csv('finance_economics_dataset.csv')
fin
```

```python
# Q.(1) What is the shape of the dataset?
print(fin.shape)
```

```python
# Q.(2) What are the column names and their data types?
print(fin.dtypes)
```

```python
# Q.(3) How many unique stock indices are there?
unique_indices = fin['Stock Index'].nunique()
index_names = fin['Stock Index'].unique()

print(f"Number of unique stock indices: {unique_indices}")
print(f"Indices: {index_names}")
```

```python
# Q.(4) What is the date range of the dataset?
fin['Date'] = pd.to_datetime(fin['Date'])
min_date = fin['Date'].min()
max_date = fin['Date'].max()

print(f"Date Range: {min_date.strftime('%Y-%m-%d')} to {max_date.strftime('%Y-%m-%d')}")
```

```python
# Q.(5) Are there any missing values?
print(fin.isnull().sum())
```

```python
# Q.(6) Are there negative values in columns that should be non-negative?
numerical_cols = ['Open Price', 'Close Price', 'Daily High', 'Daily Low', 'Trading Volume']
print(fin[numerical_cols].min())
```

```python
# Q.(7) What is the summary of GDP Growth (%)?
print(fin['GDP Growth (%)'].describe())
```

```python
# Q.(8) Are there rows with zero or near-zero trading volume?
zero_volume_count = (fin['Trading Volume'] <= 0).sum()
print(f"Rows with zero/negative volume: {zero_volume_count}")
print(f"Minimum trading volume recorded: {fin['Trading Volume'].min()}")
```

```python
# Q.(9) Are there any duplicate rows?
duplicate_count = fin.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Q.(10)Are there outliers in GDP, Gold, or Oil prices?
for keyword in ['gdp', 'gold', 'oil']:
    col = [c for c in fin.columns if keyword.lower() in c.lower()][0]
    Q1 = fin[col].quantile(0.25)
    Q3 = fin[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = fin[(fin[col] < (Q1 - 1.5 * IQR)) | (fin[col] > (Q3 + 1.5 * IQR))]
    print(f"Column '{col}' has {len(outliers)} outliers.")
```

```python
# Q.(11) What is the summary of Inflation Rate (%)?
inf_col = [c for c in fin.columns if 'inflation' in c.lower()][0]
print(fin[inf_col].describe())
```

```python
# Q.(12) What is the average unemployment rate?
unemp_col = [c for c in fin.columns if 'unemployment' in c.lower()][0]
print("Average Unemployment Rate:", fin[unemp_col].mean())
```

```python
# Q.(13) Which index has the highest trading volume?
idx_col = [c for c in fin.columns if 'index' in c.lower()][0]
vol_col = [c for c in fin.columns if 'volume' in c.lower()][0]
print("Index with highest total trading volume:", fin.groupby(idx_col)[vol_col].sum().idxmax())
```

```python
# Q.(14) How many stock records are from each index?
idx_col = [c for c in fin.columns if 'index' in c.lower()][0]
print(fin[idx_col].value_counts())
```

```python
# Q.(15) What is the correlation between inflation and interest rate?
inf_col = [c for c in fin.columns if 'inflation' in c.lower()][0]
int_col = [c for c in fin.columns if 'interest' in c.lower()][0]
print("Correlation:", fin[[inf_col, int_col]].corr().iloc[0, 1])
```

```python
# Q.(16) What is the average Consumer Confidence Index?
cci_col = [c for c in fin.columns if 'consumer confidence' in c.lower()][0]
print("Average Consumer Confidence Index:", fin[cci_col].mean())
```

```python
# Q.(17) Which column has the highest standard deviation?
numeric_cols = fin.select_dtypes(include=['int64', 'float64'])
print("Column with highest standard deviation:", numeric_cols.std().idxmax())
```

```python
# Q.(18) What is the highest gold price recorded?
gold_col = [c for c in fin.columns if 'gold' in c.lower()][0]
print("Highest Gold Price:", fin[gold_col].max())
```

```python
# Q.(19) Which date had the highest crude oil price?
oil_col = [c for c in fin.columns if 'oil' in c.lower()][0]
date_col = [c for c in fin.columns if 'date' in c.lower()][0]
print("Date with highest crude oil price:", fin.loc[fin[oil_col].idxmax(), date_col])
```

```python
# Q.(20) What is the average corporate profit?
profit_col = [c for c in fin.columns if 'profit' in c.lower()][0]
print("Average Corporate Profit:", fin[profit_col].mean())
```

```python
# Q.(5) Convert Order Date and Ship Date to datetime.
import pandas as pd

# 1. Reload a fresh copy of the dataset to reset everything
rtl = pd.read_csv('Retail Data.csv')

# 2. Run Q5: Convert dates to datetime (Run ONLY ONCE)
rtl['Order Date'] = pd.to_datetime(rtl['Order Date'], format='%d-%m-%Y')
rtl['Ship Date'] = pd.to_datetime(rtl['Ship Date'], format='%d-%m-%Y')
print("--- New Q5 Verification ---")
print(rtl[['Order Date', 'Ship Date']].dtypes)
print(rtl[['Order Date', 'Ship Date']].head(2))
```

```python
# Q.(6) Check for future or inconsistent shipping dates.
inconsistent = (rtl['Ship Date'] < rtl['Order Date']).sum()
print("Inconsistent dates:", inconsistent)
```

```python
# Q.(7) Convert price columns to numeric (remove $ and commas).

price_cols = ['Cost Price', 'Retail Price', 'Profit Margin', 'Sub Total', 'Discount $', 'Order Total', 'Shipping Cost', 'Total']
for col in price_cols:
    rtl[col] = rtl[col].astype(str).str.replace('$', '').str.replace(',', '')
    rtl[col] = pd.to_numeric(rtl[col])

# Clean the percentage sign and divide by 100
rtl['Discount %'] = pd.to_numeric(rtl['Discount %'].astype(str).str.replace('%', '')) / 100
print("\n--- New Q7 Verification ---")
print(rtl[['Retail Price', 'Total', 'Discount %']].dtypes)
print(rtl[['Retail Price', 'Total', 'Discount %']].head(2))
```

```python
# Q.(12) Create a new column for shipping duration.
rtl['Shipping Duration'] = (rtl['Ship Date'] - rtl['Order Date']).dt.days
# --- Q12 Verification ---
print("Data Type:", rtl['Shipping Duration'].dtype)
print("\nFirst 5 values:")
print(rtl[['Order Date', 'Ship Date', 'Shipping Duration']].head())
```

```python
# Q.(14) Are all discount percentages matching discount dollar amounts?
rtl['Order Date'] = pd.to_datetime(rtl['Order Date'], format='%d-%m-%Y')
rtl['Ship Date'] = pd.to_datetime(rtl['Ship Date'], format='%d-%m-%Y')

price_cols = ['Cost Price', 'Retail Price', 'Profit Margin', 'Sub Total', 'Discount $', 'Order Total', 'Shipping Cost', 'Total']
for col in price_cols:
    rtl[col] = pd.to_numeric(rtl[col].astype(str).str.replace('$', '').str.replace(',', ''))
rtl['Discount %'] = pd.to_numeric(rtl['Discount %'].astype(str).str.replace('%', '')) / 100
mismatches = ((rtl['Sub Total'] * rtl['Discount %'] - rtl['Discount $']).abs() > 0.05).sum()
print("Correct Mismatched discounts:", mismatches)
```

```python
# Q.(15) Check for mismatches in total calculation’
mismatches_total = ((rtl['Order Total'] + rtl['Shipping Cost'] - rtl['Total']).abs() > 0.05).sum()
print("Mismatched totals:", mismatches_total)
```
